```python
"""
折线图：趋势随时间变化
条形图/柱状图：类别之间对比
饼图：整体组成比例
散点图：两变量相关性
箱线图：数据分布、异常

by my word:
折线图：数据的趋势与变化
条形图/柱状图：同类数据之间的横向对比
饼图：整体与局部的占比关系
散点图：两个变量之间的相关性
箱线图：数据的分布、异常情况
"""
```

```python
"""
通用的步骤
"""

# 1. 导入库，设置字体
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = ['Hei']

# 2. 说明数据
month = ['1月','2月','3月', '4月', '5月', '6月', '7月', '8月']
sales = [100, 150, 80, 130, 150, 170, 200, 90]
things = ['学习','娱乐','运动','睡觉','其他']
times = [6,4,1,8,5]
x = []
y = []
data = {
    '语文':[82, 85, 88, 70, 90, 75, 84, 83, 95],
    '数学':[75, 80, 79, 93, 88, 82, 87, 89, 92],
    '英语':[70, 72, 68, 65, 78, 80, 85, 90, 95]
}

# 3. 表格的配置
plt.figure(figsize=(10,5))

# 4. 图表的选择
# 折线图
plt.plot(month, sales,)
# 柱状图 / 条形图为 plt.barh
plt.bar(month, sales,)
# 饼状图
plt.pie(times,labels=things)
# 散点图
plt.scatter(x, y)
# 箱线图
plt.boxplot(data.values(), tick_labels=data.keys())

# 5. 图表的配置
# 标题
plt.title('标题')
# 轴标
plt.xticks(rotation=0, fontsize=15)
# 轴标题
plt.ylabel('销售额（万元）',fontsize=10)
# 轴范围
plt.ylim(0,250)
# 图例
plt.legend(loc='upper left',fontsize=12)
# 浮动文字
for x,y in zip(month,sales):
    plt.text(x, y+5, str(y), ha='center', va='bottom',fontsize=10)
# 网格线
plt.grid(True, alpha=0.2, linestyle='--', color='#ff8888')
# ...

# 6. 图表的绘制
plt.tight_layout()  # 自动排版
plt.show()          # 买定离手
```
