```python
import matplotlib.pyplot as plt
import pandas as pd
# 设置字体
from matplotlib import rcParams
rcParams['font.family'] = ['Hei']

# 创建图表，设置大小
plt.figure(figsize=(8,5))

# 要绘制的数据
things = ['学习','娱乐','运动','睡觉','其他']
times = [6,4,1,8,5]

# 绘制环形图
plt.pie(times,labels=things,
        autopct='%1.1f%%', #第一个百分号是固定格式，第二个是转义，第三个是表示显示百分数
        shadow=True,
        startangle=45, # 起始角度
        wedgeprops={'width':0.5},   # 环形图宽度
        pctdistance=0.7,        # 文字距离圆心位置
        )

# 添加标题
plt.title('一日时间安排', fontsize=20)
plt.text(0, 0, '总计:24h', va='center', ha='center', fontsize=10)

# 添加坐标轴的标签

# 添加图例

# 添加网格线

# 显示图标
plt.tight_layout()  # 自动优化排版
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
# 设置字体
from matplotlib import rcParams
rcParams['font.family'] = ['Hei']

# 创建图表，设置大小
plt.figure(figsize=(8,5))

# 要绘制的数据
things = ['学习','娱乐','运动','睡觉','其他']
times = [6,4,1,8,5]
explode = [0.1, 0, 0, 0, 0]
# colors = ['#ff2222', '#ff4444', '#ff6666', '#ff8888', '#ffcccc']

# 绘制爆炸饼图
plt.pie(times,labels=things,
        autopct='%1.1f%%', #第一个百分号是固定格式，第二个是转义，第三个是表示显示百分数
        shadow=True,
        startangle=0, # 起始角度
        # colors=colors,
        explode=explode,
        )

# 添加标题
plt.title('一日时间安排', fontsize=20)

# 添加坐标轴的标签

# 添加图例

# 添加网格线

# 显示图标
plt.tight_layout()  # 自动优化排版
plt.show()
```
