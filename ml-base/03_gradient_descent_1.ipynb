```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 定义目标函数
def f(x):
    return (x + 3)**2 + 5

# 定义梯度函数（导函数）
def gradient(x):
    return 2 * x + 6

# 用列表保存点的变化轨迹
x_list = []
y_list = []

# 定义超参数 学习率 和 x 初始值
alpha = 0.1
x = -1

# 重复迭代 100 次
while abs(grad := gradient(x)) > 1e-6:
    y = f(x)
    # 记录当前点的坐标
    x_list.append(x)
    y_list.append(y)
    print(f'f(x={x:.10f}) = {f(x):.10f}')

    # 计算梯度
    # grad = gradient(x)
    # 更新参数
    x = x - alpha * grad
print(f'node_num:{len(x_list)}')
```

```python
# 画图
x = np.arange(-3.5, -0.5, 0.01)
plt.plot(x, f(x))
plt.plot(x_list, y_list, c = 'r')
plt.scatter(x_list, y_list, c = 'r')
plt.grid(True, linestyle = '--')

plt.show()
```
