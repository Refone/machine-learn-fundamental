```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns
```

定义类别标签

```python

labels = ["猫", "狗"]
# 定义数据（模拟预测值与真实值）
y_true = ["猫", "猫", "猫", "猫", "猫", "猫", "狗", "狗", "狗", "狗"]
y_pred = ["猫", "猫", "狗", "猫", "猫", "猫", "猫", "猫", "狗", "狗"]
```

得到混淆矩阵

```python
# 最后一个 label 控制显示哪些 label，以及 label 排序，这里只有 猫 和 狗，作用不大
matrix = confusion_matrix(y_true, y_pred, labels=labels)
print(matrix)
```

```python
pd.DataFrame(matrix, columns=labels, index=labels)
```

```python
sns.heatmap(matrix, annot=True, fmt="d", cmap="Greens")
plt.show()
```

准确率 $Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_true, y_pred)
print(accuracy)
```

精确率 $Precision = \frac{TP}{TP + FP}$

```python
from sklearn.metrics import precision_score
precision = precision_score(y_true, y_pred, pos_label='猫')
print(precision)
```

召回率 $Recall = \frac{TP}{TP + FN}$
> 池塘 100 条鱼，一网打了 80 只，有没有漏网之鱼？召回 = 找回

```python
from sklearn.metrics import recall_score
recall = recall_score(y_true, y_pred, pos_label='猫')
print(recall)
```

F1 分数 $F_1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}$
> 精准率和召回率的 **调和平均数**

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_true, y_pred, pos_label='猫')
print(f1)
```

综合报告

```python
from sklearn.metrics import classification_report
report = classification_report(y_true, y_pred, target_names=labels)
print(report)
```
