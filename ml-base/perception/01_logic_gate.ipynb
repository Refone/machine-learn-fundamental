$$
y =

1 & (w_1x_1 + w_2x_2 > \theta)

$$
y =
\begin{cases}
0 & (b + w_1x_1 + w_2x_2 \leq 0) \\
1 & (b + w_1x_1 + w_2x_2 > 0)
\end{cases}
$$

用这套公式，可以用完全一样的代码，仅仅是调整 $w_1$ $w_2$ $\theta$ 的值，

就可以实现 与门、与非门、或门

````python
# 实现与门标量版
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    res = x1 * w1 + x2 * w2
    if res <= theta:
        return 0
    else:
        return 1

print( AND(0, 0) )
print( AND(0, 1) )
print( AND(1, 0) )
print( AND(1, 1) )
````

````python
# 实现与门向量版
import numpy as np

def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7

    res = w @ x + b
    return 0 if res <= 0 else 1

print( AND(0, 0) )
print( AND(0, 1) )
print( AND(1, 0) )
print( AND(1, 1) )
````

````python
# 实现与非门向量版
import numpy as np

def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7

    res = w @ x + b
    return 0 if res <= 0 else 1

print( NAND(0, 0) )
print( NAND(0, 1) )
print( NAND(1, 0) )
print( NAND(1, 1) )
````

````python
# 实现或门向量版
import numpy as np

def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0

    res = w @ x + b
    return 0 if res <= 0 else 1

print( OR(0, 0) )
print( OR(0, 1) )
print( OR(1, 0) )
print( OR(1, 1) )
````

* 那怎么实现异或门呢？
* 仅靠单层感知机，是无法实现异或门的，但是可以靠多层感知机实现

```mermaid
flowchart LR
```

````python
# 异或门
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y

print( XOR(0, 0) )
print( XOR(0, 1) )
print( XOR(1, 0) )
print( XOR(1, 1) )
````
