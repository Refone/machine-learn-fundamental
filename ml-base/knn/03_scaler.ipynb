```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
X = [[2, 1], [3, 1], [1, 4], [2, 6]]

# 定义 Scaler
mm_scaler = MinMaxScaler(feature_range=(-1, 1))    # 默认是 0, 1
X_scaled = mm_scaler.fit_transform(X)
print(X_scaled)
```

```python
std_scaler = StandardScaler()
X_std_scaled = std_scaler.fit_transform(X)
print(X_std_scaled)
```

```python
# standard scaler 是把原数据转化为标准分布
# 标准差压缩到 0 ~ 1
# 相当于以下手动压缩：
import numpy as np

XX = np.array(X)
print(XX.shape)
mean = XX.mean(axis=0)
std = XX.std(axis=0)
print(mean, std)
```

```python
print((XX - mean) / std)
```
