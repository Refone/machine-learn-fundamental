```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X = [[2,1],[4,3],[1,4],[3,5]]
y = [0, 1, 0, 1]

X = np.array(X)
y = np.array(y)

# colors = ['#ffcc00' if v==0 else '#6622ff' for v in y ]
# plt.figure(figsize=(5,5))
# plt.axis('equal')
# plt.xlim(0, 6)
# plt.ylim(0, 6)
# plt.scatter(X[:,0], X[:,1], c=colors)
# plt.grid(True, linestyle='--', alpha=0.3)
# plt.show()
```

```python
knn = KNeighborsClassifier(n_neighbors=2, weights='distance') # 如果多个点打平，给每个点配置权重
knn.fit(X, y)

x = np.array([[1,2],[3, 3], [2,3], [5,5]])
x_class = knn.predict(x)
print(x_class)
```

```python
plt.figure(figsize=(4, 4))
plt.grid(True, linestyle='--', alpha=0.3)
plt.xlim(-0.5, 5.5)
plt.ylim(0, 6)

X1 = X[y==0]
X2 = X[y==1]

colors = ["#ffcc00", "#6622ff"]
plt.scatter(X1[:,0], X1[:,1], c=colors[0])
plt.scatter(X2[:,0], X2[:,1], c=colors[1])

for i,point in enumerate(x):
    x_color = colors[0] if x_class[i] == 0 else colors[1]
    plt.scatter(point[0], point[1], c=x_color, marker='x', s=50)
plt.tight_layout()
plt.show()
```
