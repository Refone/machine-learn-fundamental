```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
advertising = pd.read_csv('../../data/advertising.csv')
print(advertising.head())
print(advertising.describe())
print(advertising.shape)
```

```python
# 数据预处理
# 去掉第一列
advertising.drop(columns=['Unnamed: 0'], inplace=True)
print(advertising.head())
```

```python
# 处理空值
advertising.isna().sum()
```

```python
# 提取特征和标签（目标值）
X = advertising.drop('Sales', axis=1)
Y = advertising['Sales']
print("X:\n", X.head())
print("Y:\n", Y.head())
```

```python
# 计算皮尔逊相关系数
print(X.corrwith(Y, method='pearson'))
```

```python
# corr 的调用
print(advertising.corr(method='pearson'))
```

```python
# 看一看热力图
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(data=advertising.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
