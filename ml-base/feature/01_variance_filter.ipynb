```python
import numpy as np
```

```python
a = np.random.randn(1000)
print(np.var(a))
```

```python
# b = np.random.randn(1000) * 0.1
b = np.random.normal(5, 0.1, size=1000)
print(np.var(b))
```

```python
X = np.vstack((a, b)).T
# X = np.array([a,b])  # 错误！ 不是上下拼接，而是把 a b 塞进一个更高维的容器中
print(X)
print(X.shape)
```

```python
# 低方差过滤
from sklearn.feature_selection import VarianceThreshold
vt = VarianceThreshold(0.01)
X_filter = vt.fit_transform(X)
# print(X_filter)
print(X_filter.shape)
```
