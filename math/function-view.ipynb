```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

x = np.linspace(-np.pi, np.pi, 256)
y = x ** 4
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')

plt.show()
```

```python
# 关于向量
x = np.array([2, 5, 8])
y = np.array([1, 3, 7])
print('x + y', x+y)
print('3 * x', 3 * x)
print('x * y', x*y)
print('x @ y', x.dot(y))

# 计算范数
a = np.array([1, 1, 0])
print('l0_norm', np.linalg.norm(a, ord=0))
print('l1_norm', np.linalg.norm(a, ord=1))
print('l2_norm', np.linalg.norm(a, ord=2))
```

```python
# 矩阵求逆
C = np.array([[1, 2],[3, 5]])
C_inv = np.linalg.inv(C)
C_inv = np.around(C_inv).astype(int)
print(C)
print(C_inv)
```

```python
print('C @ C_inv\n', C_inv.dot(C))
```

```python
A = np.array([[1, 2, 3],[4, 5, 6]])
print('A\n', A)
```

```python
A_pinv = np.linalg.pinv(A)
print('A的伪逆矩阵\n', A_pinv)
```

```python
print('A @ A_pinv\n', A.dot(A_pinv))
```

```python
print(A[0, 0])
print(A[0][0])
print(A[:, 0])
```

```python
A = np.array([[1, 2, 3],[4, 5, 6]])
B = np.array([[1, 3, 7],[5, 0, 2]])
print('<A, B>\n', np.dot(A.flatten(), B.flatten()))
```

```python
np.dot([1, 2, 3, 4, 5, 6], [1, 3, 7, 5, 0, 2])
```

```python
# 克罗内克积
np.kron(A, B)
```

```python
# 梯度计算
f = np.array([1, 2, 4, 7, 11, 16])
grad = np.gradient(f)
print(grad)
```
