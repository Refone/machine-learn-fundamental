# 信息熵 (Entropy)

信息熵（Entropy）是表示随机变量不确定性的度量，设X是一个取有限个值的离散随机变量，其概率分布为：

$$P(X = x_i) = p_i, i = 1, 2, 3, ..., n$$

随机变量 X 的熵定义为：

$$H(X) = - \sum_{i=1}^{n}p_i \log(p_i)$$

在 $n = 2$ 的情况下，设 X 概率分布为

$$P(X=1)=p, P(X=0)=1-p$$

则，X 的信息熵为：

$$H(X) = -p \cdot \log p - (1-p) \cdot \log(1-p)$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 定义 x 范围，避免 x=0 （log 无定义）
x = np.linspace(1e-10, 1, 500)  # 从接近0到1
y = x * np.log(x)

# 绘图
plt.figure(figsize=(6, 4))
plt.plot(x, y, label=r'$f(x) = x \log x$', color='b')

# 添加参考线（x轴、y轴）
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')

# 标记最小值点（在 x = 1/e 处取得最小值 -1/e）
x_min = 1 / np.e
y_min = x_min * np.log(x_min)
plt.plot(x_min, y_min, 'ro', label=f'Min: ({x_min:.3f}, {y_min:.3f})')

# 设置坐标范围和标签
plt.xlim(0, 1)
plt.ylim(-0.4, 0.05)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'$f(x) = x \log x$ on (0, 1)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

plt.show()
```
