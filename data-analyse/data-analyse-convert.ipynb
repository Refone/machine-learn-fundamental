# 关于类型转换（提效）

```python
import pandas as pd
```

```python
# 数据类型的转换
df = pd.read_csv('../data/sleep.csv')
df.tail()
```

```python
df.dtypes
```

```python
df.dtypes['age']
```

```python
df['age'] = df['age'].astype('int8')
df.dtypes['age']
```

```python
df.gender.tail()
```

```python
df.dtypes['gender']
```

```python
df['gender'] = df['gender'].astype('category')
df.dtypes
```

```python
df.gender.tail()
```

```python
df['is_male'] = df['gender'].map({'Female':False, 'Male':True})
df.is_male.tail()
```
