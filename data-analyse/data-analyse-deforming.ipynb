# 数据变形

```python
import pandas as pd
data = {
    'ID' : [1, 2],
    'name' : ['alice', 'bob'],
    'Math' : [90, 85],
    'English' : [88, 92],
    'Science' : [95, 89]
}
df = pd.DataFrame(data)
df
```

```python
# 宽表转换成长表
df2 = pd.melt(df, id_vars=['ID', 'name'],
            var_name='Subject', value_name='Score'
            , value_vars=['Math', 'English', 'Science']
            )
df2
```

```python
# 长表转宽表
pd.pivot(df2, index=['ID', 'name'], columns=['Subject'], values='Score')
```
