# 案例
## 学生成绩统计

```python
np.random.seed(42)
scores = pd.Series(np.random.randint(50, 101, 10), index=['学生'+str(i) for i in range(1, 11)])
scores
```

```python
# 平均分
print('平均分', scores.mean())
print('最高分', scores.max())
print('最低分', scores.min())
# 高于平均分的人数
mean = scores.mean()
print('高于平均分的人数', len(scores[scores>mean]))
print('高于平均分的人数', scores[scores>mean].count())
```

```python
# 温度数据分析
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33],
                         index = ['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
print(temperatures)
print('温度超过30',temperatures[temperatures>30].count())
print('平均温度', temperatures.mean().round(2))
print('温度从高到低排序\n', temperatures.sort_values(ascending=False))
```

```python
# 找出温度变化最大的两天
diff = temperatures.diff().abs()
diff
```

```python
diff.sort_values(ascending=False)
```

```python
print('温度变化最大的两天', diff.sort_values(ascending=False).keys()[:2])
print('温度变化最大的两天', diff.sort_values(ascending=False).keys()[:2].tolist())
print('温度变化最大的两天', *(diff.sort_values(ascending=False).keys()[:2].tolist()))
```

# 案例：股票价格分析

```python
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2], index = pd.date_range('2023-01-01', periods=10))
prices
```

```python
# 计算每日收益率（当日收盘价/前日收盘价 - 1)
a = prices.pct_change()
print('每日收益率', a)
```

```python
# 找出收益率最高和最低的日期
print('收益率最高的日期', a.idxmax())
print('收益率最低的日期', a.idxmin())
```

```python
# 计算波动率 (收益率的标准差）
print(a.std())
```

# 小时销售额分析

```python
np.random.seed(42)
hours_sales = pd.Series(np.random.randint(0, 100, 24),
                         index=pd.date_range('2025-01-01', periods=24,freq='h'))
hours_sales
```

```python
# 按天重采样
day_sales = hours_sales.resample('D').sum()
day_sales
```

```python
print('营业时间销售额\n', hours_sales.between_time('8:00', '22:00'))
```

```python
day_sales = hours_sales[(hours_sales.index.hour>=8) & (hours_sales.index.hour<=22)]
print('营业时间销售额\n', day_sales)
```

```python
day_sales.sum()
```

```python
day_sales.sum() / hours_sales.sum()
```

```python
np.random.seed(42)
h = pd.Series(np.random.randint(0, 100, 24),
                         index=pd.date_range('2025-01-01', periods=24,freq='h'))
mask = (h.index.hour >= 8) & (h.index.hour <= 22)
b = h[mask]
n_b = h[~mask]
print(b.sum()/n_b.sum())
```

```python
print(h.nlargest(3))
```
