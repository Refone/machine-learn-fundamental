```python
# 1. 导入必要的库
import pandas as pd
import numpy as np

# 2. 导入数据
df = pd.read_csv('../data/penguins.csv')
print(df.info())
df.head()
```

```python
# 3. 数据清洗
# 处理缺失数据
# 检查处理完毕
df2 = df
print(df2.isna().sum())
print('total rows:', df2.shape[0])
df2 = df2.dropna()
print(df2.isna().sum())
print('total rows:', df2.shape[0])
```

```python
# 4. 数据特征构造
# 性别转分类
df3 = df2
print(df3['sex'].dtype)
df3['sex'] = df3['sex'].astype('category')
print(df3['sex'].dtype)
# 新增 bill_ratio len / dept
df3['bill_ratio'] = (df3['bill_length_mm'] / df3['bill_depth_mm']).round(3)
df3
```

```python
# 5. 数据分析
# 体重，数据分箱，高 中 低 mass_level
df4 = df3
df4['mass_level'] = pd.cut(df4['body_mass_g'],
                           bins=[0,2000,4000,np.inf],
                           labels=['light','middle','heavy'])
# 按性别分组，按岛屿分组，agg 分别看一看 均值和数量
df4.groupby(['sex','island']).agg({
    'body_mass_g':['mean','count']
})
```
