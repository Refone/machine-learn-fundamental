```python
import matplotlib.pyplot as plt
import random
import math
# 设置字体
from matplotlib import rcParams
rcParams['font.family'] = ['Hei']

# 创建图表，设置大小
plt.figure(figsize=(8,5))

# 要绘制的数据
x = []
y = []
for i in range(500):
    tx = random.uniform(1,10)
    ty = ((math.sqrt(random.random())) + 0.5) * tx
    # print(tx,ty)
    x.append(tx)
    y.append(ty)

# 绘制折线图
plt.scatter(x, y,
            color='#ffcc00',
            alpha=0.5,
            s=20,   # 原点大小
            )

# 添加标题
plt.title('Scatter Plot')

# 添加网格线
plt.grid(True, alpha=0.2, linestyle='--', color='#ff8888')

# 绘制回归线
plt.plot([1, 10], [1, 14],
         linestyle='-.',
         linewidth=2,
         color='#e53d30',
         )

# 显示图标
plt.tight_layout()
plt.show()
```
