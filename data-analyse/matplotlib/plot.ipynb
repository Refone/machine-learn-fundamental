```python
import matplotlib.pyplot as plt
import pandas as pd
# 设置字体
from matplotlib import rcParams
rcParams['font.family'] = ['Hei']

# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 要绘制的数据
month = ['1月','2月','3月', '4月', '5月', '6月', '7月', '8月']
sales = [100, 150, 80, 130, 150, 170, 200, 90]

# 绘制折线图
plt.plot(month, sales,
         color='#ff8888',
         linestyle='--',
         linewidth=2,
         marker='o',
         markersize=8,
         label='产品A')

# 添加标题
plt.title('2025年销售趋势', fontsize=20, color='#ff8888')

# 添加坐标轴的标签
plt.xticks(rotation=0, fontsize=15)
plt.yticks(rotation=20,fontsize=12)
plt.ylabel('销售额（万元）',fontsize=10)
# plt.xlabel('月份', fontsize=15)
plt.ylim(0,250)

# 添加图例
plt.legend(loc='upper left',fontsize=12)
for x,y in zip(month,sales):
    plt.text(x, y+5, str(y), ha='center', va='bottom',fontsize=10)

# 添加网格线
plt.grid(True, alpha=0.2, linestyle='--', color='#ff8888')

# 显示图标
plt.show()
```
