```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'Hei'

month = ['1', '2', '3', '4', '5']
sales = [100, 200, 80, 130, 120]

# 2x2 图表中第一个
f1 = plt.subplot(2, 2, 1)
f1.plot(month, sales)

# 2x2 图表中第二个
f2 = plt.subplot(2, 2, 2)
f2.barh(month, sales)
```
