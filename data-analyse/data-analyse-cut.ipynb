# 数据分箱

```python
# pd.cut(x, bins, labels)
import pandas as pd

df = pd.read_csv('../data/employees.csv')
df.head(10)
```

```python
df1 = df.head(10)[['employee_id', 'salary']]
df1
```

```python
pd.cut(df1['salary'], bins = 2)
```

```python
pd.cut(df1['salary'], bins = 2).value_counts()
```

```python
pd.cut(df['salary'], bins = [0, 10000, 20000, 30000]).value_counts()
```

```python
pd.cut(df1['salary'], bins = [0, 10000, 20000, 30000], labels=['low','middle', 'high']).value_counts()
```

```python
df1['收入范围'] = pd.cut(df1['salary'], bins = [0, 10000, 20000, 30000], labels=['low','middle', 'high'])
df1
```

```python
pd.qcut(df1['salary'],3).value_counts()
```

```python
# 睡眠数据
df = pd.read_csv('../data/sleep.csv')
df.head(10)
df1 = df.head(10)[['person_id', 'sleep_quality']]
df1
```

```python
pd.cut(df['sleep_quality'], bins = 3).value_counts()
```

```python
pd.qcut(df['sleep_quality'], 3).value_counts()
```

```python
df1['睡眠质量'] = pd.cut(df['sleep_quality'], bins = 3, labels=['差','中', '优'])
df1
```

```python
df1['睡眠质量'].value_counts()
```

```python
# 分箱的逻辑是给数据分类
print(df1['睡眠质量'].dtypes)
```
