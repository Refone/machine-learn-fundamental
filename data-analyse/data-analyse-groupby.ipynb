```python
import numpy as np
# 分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数()

import pandas as pd
df = pd.read_csv('../data/employees.csv')
df
```

```python
df['department_id'] = df['department_id'].fillna(0).astype('int64')
df
```

```python
# 计算不同部门的平均薪资
df.groupby('department_id').groups
```

```python
df.groupby('department_id').get_group(20)
df2 = df.groupby('department_id')[['salary']].mean().round(0) # 先分组，后聚合
df2 = df2.reset_index()
df2.sort_values('salary', ascending=False)
```

```python
df3 = df.groupby(['department_id', 'job_id'])[['salary']].mean().round(0)
df3
```
