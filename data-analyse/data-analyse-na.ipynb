```python
# 缺失值的处理
import pandas as pd
import numpy as np

s = pd.Series([1, 2, np.nan, None, pd.NA])
print(s)
```

```python
# 查看是否缺失值
s.isna()
```

```python
s.isnull()
```

```python
df = pd.DataFrame([[1,pd.NA, 2], [2, 3, 5], [None, 4, 6]],columns=['第1列','第2列','第3列'])
df
```

```python
df.isna()
```

```python
df.isna().sum(axis=1)
```

```python
df.isna().sum()
```

```python
# 剔除缺失值
print(s)
print('-'*50)
print(s.dropna())
```

```python
print(df)
print('-'*50)
print(df.dropna())

print('how=all', '-'*50)
print(df.dropna(how='all'))
```

```python
print('thresh', '-'*50)
print(df.dropna(thresh=3))

print('vertical', '-'*50)
print(df.dropna(axis=1))
```

```python
print(df.dropna(subset=['第1列']))
```

```python
#  填充缺失值
df = pd.read_csv('../data/weather_withna.csv')
df.head()
```

```python
df.isna().sum(axis=0)
```

```python
df.head()
print(df.fillna({'temp_max':20}).tail())   # 使用字典来填充
print(df.fillna(df[['wind']].mean()).tail())    # 用统计值填充
```

```python
print(df.ffill().tail()) # 用前驱填充（传感器信号丢失）
```

```python
print(df.bfill().tail()) # 用后继填充
```
