```python
# 时间的处理
import pandas as pd
d = pd.Timestamp('2015-05-02 10:22')
d
```

```python
print(type(d))
```

```python
print('年', d.year)
print('月', d.month)
print('日', d.day)
print(d.hour, d.minute, d.second)
print('季度', d.quarter)
print('是否 d.is_xxx')
print('是否闰年', d.is_leap_year)
```

```python
print('星期几', d.day_name())
print('转换为天',d.to_period('D'))
print('转换为季度:', d.to_period('Q'))
print('年度', d.to_period('Y'))
print('月度', d.to_period('M'))
print('周维度', d.to_period('W'))
```
