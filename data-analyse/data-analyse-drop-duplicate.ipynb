# 关于重复数据

```python
import pandas as pd
data = {
    "name" : ['alice', 'alice', 'bob', 'alice', 'jack', 'bob'],
    'age' : [26, 25, 30, 25, 35, 30],
    'city' : ['NY', 'NY', 'LA', 'NY', 'SF', 'LA']
}
df = pd.DataFrame(data)
df
```

```python
df.duplicated() # 一整条记录都是一样的，标记为重复，返回 True
```

```python
df.drop_duplicates()
```

```python
df.drop_duplicates(subset=['name']) # 根据指定列去重
```

```python
df.drop_duplicates(subset=['name'], keep='last') # 保留最后（最新）者
```
