```python
import numpy as np
temps = np.array([28, 30, 29, 31, 32, 30, 29])
print('平均气温', np.mean(temps).round(2))
print('最高气温', np.max(temps).round(2))
print('最低气温', np.min(temps).round(2))

print('超过 30 度的天数', len(temps[temps>30]))
print('超过 30 度的天数', np.cumsum(np.where(temps>30, 1, 0))[-1])
print('超过 30 度的天数', np.count_nonzero(np.where(temps>30)))
```

```python
scores = np.array([85, 90, 78, 92, 88])
print(scores)
print('平均分', np.mean(scores).round(1))
print('中位数', np.median(scores).round(1))
print('标准差', np.std(scores).round(2))
print('10 分制', (scores/10).round(1))
```

```python
A = np.array([[1, 2],[3, 4]])
B = np.array([[5, 6], [7, 8]])
print('A+B\n',A+B)
print('A*B\n',A*B)
print('A@B\n',A@B)
print('A@B\n',np.dot(A,B))
```

```python
np.random.seed(0)
arr = np.random.randint(0, 10, (3, 4))
print(arr)
```

```python
print('每列最大值', np.max(arr, axis=0))
print('每行最小值', np.min(arr, axis=1))
```

```python
print('每行奇数替换为-1\n', np.where(arr % 2 != 0, -1, arr))
arr[arr%2==1]=-1
print(arr)
```

```python
# 第 5 题
np.random.seed(0)
arr = np.random.randint(0, 20, (5, 5))
print(arr)
```

```python
# 第 6 题
sales = np.array([120, 135, 110, 125, 130, 140])
print('总额', np.sum(sales))
print('方差', np.var(sales).round(1))
print('均值', np.mean(sales).round(1))
```

```python
print('最高月份', np.argmax(sales)+1)
print('最低月份', np.argmin(sales)+1)
```

```python
# 20.
sale = np.array([20, 25, 22, 30, 28])
cost = np.array([15, 18, 16, 22, 20])

profit = sale - cost
print('每天利润', profit)
print('利润平均值', np.mean(profit).round(1), '标准差', np.std(profit).round(1))
print('利润最高的天数', len(profit[profit == np.max(profit)]))
```
